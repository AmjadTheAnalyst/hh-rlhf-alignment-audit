## Research questions from the revised proposal

- **RQ1.** How often does the chosen reply gain on one coded value while losing on another, and which directed trade-offs recur within the helpfulness and harmlessness subsets?
- **RQ2.** Among pairs with independently checkable factual claims, how does the rate of unsupported high-confidence claims differ between chosen and rejected replies, and does the gap vary by response strategy?
- **RQ3.** Which identifiable stakeholder interests are addressed by each reply, and how often does a decision-relevant condition about those interests remain unspecified in the shared prompt?
- **RQ4.** When one missing condition is supplied in two plausible ways, how often does the audit judgment change between preference for A, preference for B, tie and indeterminate?
- **RQ5.** On a held-out set, how many independently adjudicated review cases does a rule using claim evidence and missing context capture at a matched review budget compared with a refusal-only rule?

## Study protocol

1. Download, pin and hash the official helpful-base and harmless-base test/train archives. Split each dialogue at the last assistant reply. Reject mismatched prompts and identical replies, then deduplicate normalized prompts.
2. Sample 250 test pairs per subset and reserve 100 per subset for held-out evaluation. Use a separate 40-pair pilot from the training archives. Before coding, audit near-duplicate prompts across splits and freeze any group or eligibility decisions. The automated filter handles normalized exact prompts.
3. Randomize A/B position. Pilot the rubric, freeze conventions and blind both reviewers to the original choice. Double-code the pilot and at least 125 main pairs. Record agreement before adjudication.
4. Code usefulness, safety and epistemic care. Independently verify checkable consequential claims and code each applicable stakeholder's visibility and one primary missing condition. RQ1–RQ3 use adjudicated records.
5. Draw up to 80 eligible RQ4 cases across subsets. Validate that two plausible conditions differ on one fact, keep both replies fixed, randomize presentation, and record A/B/tie/indeterminate judgments under each condition.
6. Independently adjudicate every held-out pair while blinded to cue codes, ranks and original preference. Compare refusal-only and evidence-plus-context screening at the same 20% budget. Use identical deterministic tie handling.
7. Report subgroup denominators, paired uncertainty, unresolved cases and sensitivity checks. The notebook displays completed exhibits, exports vector PDFs and CSVs, and packages aggregate outputs and provenance in one ZIP.

**Interpretation.** The proposal contains 20 dummy exhibits to illustrate reporting. This notebook exports 10 figures and 10 tables from supplied study observations. A run-status manifest records missing inputs and the resulting coverage.

**Critical distinction:** HH-RLHF provides conversation pairs and preferences. Claims, stakeholder roles, counterfactual judgments and independent review outcomes require original annotation. The automated notebook performs the computation after those scientific inputs are collected.

### 1 · Set paths and study constants

Configure all five questions. To freeze a publication run, set REVISION to the immutable Hugging Face commit SHA and preserve the source hashes.

```python
from pathlib import Path
from urllib.request import urlopen
import gzip, hashlib, json, re, shutil, warnings, zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, FileLink

WORK = Path("/kaggle/working")
RAW = Path("/kaggle/input/hh-rlhf-raw")
ANNOTATIONS = Path("/kaggle/input/hh-rlhf-annotations")
REVISION = ""  # Blank resolves the current immutable SHA and records it.
SEED, BOOT = 140026, 2000
N_MAIN, N_PILOT, N_HOLD, N_PROBE = 250, 20, 100, 80  # First three counts per subset.
SUBSETS = ["helpful-base", "harmless-base"]
DIMS = ["usefulness", "safety", "epistemic_care"]
RESULTS, PRIVATE = WORK/"hh_rlhf_results", WORK/"hh_rlhf_private"
ZIP = WORK/"hh_rlhf_results.zip"
COLORS = ["#176281", "#B65335", "#45866E", "#80558C", "#9CA9B2"]
if REVISION and not re.fullmatch(r"[0-9a-f]{40}", REVISION):
    raise ValueError("REVISION must be an immutable 40-character commit SHA.")
WORK.mkdir(parents=True,exist_ok=True)
for p in (RESULTS,PRIVATE):
    if p.exists(): shutil.rmtree(p)
    p.mkdir()
ZIP.unlink(missing_ok=True)
plt.rcParams.update({"font.family":"DejaVu Sans","pdf.fonttype":42,"figure.dpi":130,
                     "axes.spines.top":False,"axes.spines.right":False})
provenance, errors, status = [], [], {}
def rng(step): return np.random.default_rng(SEED + step)
def identity(s): return hashlib.sha256(s.encode()).hexdigest()[:24]
```

### 2 · Acquire and parse the official corpus

Apply the same eligibility rules to both strata. A source acquisition failure is reported and defers the dependent questions.

```python
speaker = re.compile(r"(?:\A|\n\n)(Human|Assistant):")
def norm(s): return " ".join(s.casefold().split())
def last_turn(s):
    hits = list(speaker.finditer(s))
    if not hits or hits[-1].group(1)!="Assistant" or not any(h.group(1)=="Human" for h in hits[:-1]):
        raise ValueError("malformed dialogue")
    m=hits[-1]
    p,r=s[:m.start()].strip(),s[m.end():].strip()
    if not p or not r: raise ValueError("empty dialogue")
    return p,r

rev=REVISION
local_only=all((RAW/a/f"{b}.jsonl.gz").exists() for a in SUBSETS for b in ("train","test"))
if not rev and not local_only:
    try:
        with urlopen("https://huggingface.co/api/datasets/Anthropic/hh-rlhf",timeout=30) as f:
            rev=json.load(f)["sha"]
    except (OSError,ValueError,KeyError) as e:
        errors.append(f"Could not resolve dataset revision: {e}")
records=[]
if not errors:
    for subset in SUBSETS:
        for split in ("train","test"):
            path=RAW/subset/f"{split}.jsonl.gz"
            source="attached archive"
            try:
                if not path.exists():
                    source=f"https://huggingface.co/datasets/Anthropic/hh-rlhf/resolve/{rev}/{subset}/{split}.jsonl.gz"
                    path=PRIVATE/"downloads"/subset/f"{split}.jsonl.gz"
                    path.parent.mkdir(parents=True,exist_ok=True)
                    with urlopen(source,timeout=90) as f,path.open("wb") as g: shutil.copyfileobj(f,g)
                with path.open("rb") as f: digest=hashlib.file_digest(f,"sha256").hexdigest()
                eligible=excluded=0
                with gzip.open(path,"rt",encoding="utf8") as f:
                    for line_id,line in enumerate(f,1):
                        try:
                            row=json.loads(line)
                            p,c=last_turn(row["chosen"])
                            p2,r=last_turn(row["rejected"])
                            if p!=p2 or norm(c)==norm(r): raise ValueError("ineligible pair")
                        except (ValueError,TypeError,KeyError):
                            excluded+=1
                            continue
                        records.append((identity(json.dumps([subset,split,p,c,r])),identity(norm(p)),
                                        subset,split,line_id,p,c,r))
                        eligible+=1
                provenance.append(dict(subset=subset,split=split,source=source,eligible=eligible,
                                       excluded=excluded,sha256=digest,revision=rev or "attached revision not declared"))
            except (OSError,ValueError) as e:
                errors.append(f"{subset}/{split}: {e}")
raw=pd.DataFrame(records,columns=["pair_id","cluster_id","subset","source_split","line","prompt","chosen","rejected"])
if errors: print(*errors,sep="\n")
display(pd.DataFrame(provenance))
```

### 3 · Freeze sample and blinded assignments

Create pilot, primary, secondary, and independent held-out packets. The private key maps original preference only after coding. This applies to RQ1–RQ5.

```python
pairs,pilot=pd.DataFrame(),pd.DataFrame()
double_ids=set()
if not errors and len(raw):
    canonical=(raw.assign(priority=raw.source_split.map({"test":0,"train":1}))
               .sort_values(["priority","pair_id"]).drop_duplicates("cluster_id"))
    blocks={"test":[],"train":[]}
    random=rng(1)
    for subset in SUBSETS:
        for split,n in [("test",N_MAIN),("train",N_PILOT)]:
            x=canonical.loc[(canonical.subset==subset)&(canonical.source_split==split)]
            if len(x)<n:
                errors.append(f"{subset}/{split}: {len(x)} eligible after dedup; {n} required")
                continue
            x=x.iloc[random.permutation(len(x))[:n]].copy()
            x["partition"]="pilot" if split=="train" else "development"
            if split=="test": x.loc[x.index[:N_HOLD],"partition"]="heldout"
            x["chosen_side"]=random.choice(["A","B"],len(x))
            x["reply_a"]=np.where(x.chosen_side.eq("A"),x.chosen,x.rejected)
            x["reply_b"]=np.where(x.chosen_side.eq("B"),x.chosen,x.rejected)
            blocks[split].append(x)
    if not errors:
        pairs=pd.concat(blocks["test"],ignore_index=True)
        pilot=pd.concat(blocks["train"],ignore_index=True)
        key=["pair_id","cluster_id","subset","source_split","line","partition","chosen_side"]
        pd.concat([pairs,pilot])[key].to_csv(PRIVATE/"private_sample_key.csv",index=False)
        fields=["pair_id","prompt","reply_a","reply_b"]
        double=pd.concat([g.sample(n=int(np.ceil(len(g)*.25)),random_state=SEED+i)
                          for i,(_,g) in enumerate(pairs.groupby(["subset","partition"]))])
        double_ids=set(double.pair_id)
        for name,block in [("main",pairs),("pilot",pilot),("secondary",double),
                           ("independent_heldout",pairs.loc[pairs.partition.eq("heldout")])]:
            block[fields].sample(frac=1,random_state=SEED).to_csv(PRIVATE/f"{name}_packet.csv",index=False)
        display(pairs.groupby(["subset","partition"]).size().unstack())
if pairs.empty: print("Frozen sample unavailable.",*errors,sep="\n")
```

### Private annotation schema

- **values.csv:** one row per pair and side, with 0/1/2 ordinal scores (deficient/mixed/strong) or U (unrateable). Higher means greater usefulness, safety or epistemic care. Strategy is direct, qualified or refusal.
- **reliability.csv:** independent pre-adjudication primary/secondary scores for the pilot and allocated main pairs. Fill before discussing disagreements.
- **coverage.csv:** inventory per reply: complete (one or more claim rows), no_claim or unassessed. Claims include evidence status supported, contradicted, insufficient or uncheckable. Use the same documented search procedure and effort cap across replies. Save source URL and date for supported/contradicted claims. All claims require a private rationale.
- **roles.csv:** one row per applicable role instance (including multiple individuals of one role type). Scope roles to those identifiable in the shared conversation. Code visibility in A and B as explicit, implied, absent or unresolved. **context.csv** inventories one primary missing condition, or no/unresolved, for every pair.
- **probes.csv:** one row per drawn pair. Record both values of the condition and independent validation. Judgments A/B/tie/indeterminate refer to formal condition IDs rather than presentation order. Retain failed checks and exclude them from paired transitions.
- **reviews.csv:** independent, blinded adjudications on all held-out pairs. A yes means the pair needs manual review before routine reuse because a material ethical issue would be obscured or an ambiguity needs clarification. A reviewer provides a reason for yes, no or unresolved.

### 4 · Validate independent and adjudicated coding inputs

Write blank private templates. Validate identifiers, completeness, categories and nonempty rationales so estimates depend on recorded human evidence.

```python
spec={
"values":"pair_id side usefulness safety epistemic_care strategy",
"reliability":"pair_id side reviewer usefulness safety epistemic_care",
"coverage":"pair_id side inventory",
"claims":"pair_id side claim_id claim_text confidence consequential evidence source_url access_date rationale",
"roles":"pair_id role_id role_type visibility_A visibility_B",
"context":"pair_id missing condition_type condition_note",
"probes":"pair_id condition_1 condition_2 order judgment_1 judgment_2 validated rationale",
"reviews":"pair_id needs_review rationale"}
key={"values":2,"reliability":3,"coverage":2,"claims":3,"roles":2,"context":1,"probes":1,"reviews":1}
states=["A","B","tie","indeterminate"]
visibility=["explicit","implied","absent","unresolved"]
categories={
"values":{**{d:["0","1","2","U"] for d in DIMS},"strategy":["direct","qualified","refusal"]},
"reliability":{**{d:["0","1","2","U"] for d in DIMS},"reviewer":["primary","secondary"]},
"coverage":{"inventory":["complete","no_claim","unassessed"]},
"claims":{"confidence":["categorical","qualified"],"consequential":["yes","no"],
          "evidence":["supported","contradicted","insufficient","uncheckable"]},
"roles":{"role_type":["user","named_third_party","other_affected_group"],
         "visibility_A":visibility,"visibility_B":visibility},
"context":{"missing":["yes","no","unresolved"]},
"probes":{"order":["1_first","2_first"],"judgment_1":states,"judgment_2":states,"validated":["yes","no"]},
"reviews":{"needs_review":["yes","no","unresolved"]}}
data, pending = {}, {}
main=set(pairs.pair_id) if len(pairs) else set()
pilots=set(pilot.pair_id) if len(pilot) else set()
held=set(pairs.loc[pairs.partition.eq("heldout"),"pair_id"]) if len(pairs) else set()
for name,schema in spec.items():
    cols=schema.split()
    pd.DataFrame(columns=cols).to_csv(PRIVATE/f"{name}.csv",index=False)
    try:
        d=pd.read_csv(ANNOTATIONS/f"{name}.csv",dtype=str,keep_default_na=False)
        if not set(cols)<=set(d): raise ValueError("missing required columns")
        d=d[cols].apply(lambda x:x.str.strip())
        if d.empty and name!="claims": raise ValueError("no observations")
        ids=held if name=="reviews" else pilots|double_ids if name=="reliability" else main
        if not set(d.pair_id)<=ids: raise ValueError("pair IDs do not match the frozen assignment")
        if d.duplicated(cols[:key[name]]).any() or d[cols[:key[name]]].eq("").any().any():
            raise ValueError("duplicate or blank keys")
        valid={**categories[name],**({"side":["A","B"]} if "side" in d else {})}
        if any(not d[c].isin(v).all() for c,v in valid.items()): raise ValueError("invalid category")
        if name in ["values","coverage"] and len(d)!=2*len(main): raise ValueError("incomplete reply coverage")
        if name in ["context","reviews"] and set(d.pair_id)!=ids: raise ValueError("incomplete pair inventory")
        if "rationale" in d and d.rationale.eq("").any(): raise ValueError("missing rationale")
        if name=="claims":
            verified=d.evidence.isin(["supported","contradicted"])
            if d.claim_text.eq("").any() or d.loc[verified,["source_url","access_date"]].eq("").any().any():
                raise ValueError("missing claim text or evidence citation")
        if name=="context":
            yes=d.missing.eq("yes")
            if d.loc[yes,"condition_note"].eq("").any() or not d.loc[yes,"condition_type"].isin(
                ["consent_authorization","vulnerability_exposure","other"]).all():
                raise ValueError("missing condition classification or note")
        data[name]=d
    except (OSError,ValueError) as e:
        pending[name]=str(e)
display(pd.DataFrame([{"input":n,"status":pending.get(n,"validated")} for n in spec]))
```

### 5 · Compute reliability and freeze the RQ4 experiment

Report raw agreement and weighted κ before adjudication. Draw eligible cases in subset-stratified random order and validate the returned experiment against assigned IDs and presentation order.

```python
if "reliability" in data:
    rows=[]
    for scope,ids in [("pilot",pilots),("main",double_ids)]:
        d=data["reliability"].loc[lambda x:x.pair_id.isin(ids)]
        for dim in DIMS:
            wide=d.pivot(index=["pair_id","side"],columns="reviewer",values=dim)
            wide=wide.reindex(columns=["primary","secondary"]).dropna()
            numeric=wide.loc[wide.ne("U").all(axis=1)].astype(int)
            k=np.nan
            if len(numeric):
                c=np.zeros((3,3))
                np.add.at(c,(numeric.primary.to_numpy(),numeric.secondary.to_numpy()),1)
                w=(np.arange(3)[:,None]-np.arange(3)[None,:])**2
                expected=(np.outer(c.sum(1),c.sum(0))/c.sum()*w).sum()
                if expected: k=1-(c*w).sum()/expected
            rows.append([scope,dim,2*len(ids),len(wide),len(numeric),
                         wide.primary.eq(wide.secondary).mean(),k])
    diag=pd.DataFrame(rows,columns=["stage","dimension","expected_replies","double_coded_replies",
                                   "rateable_replies","raw_agreement","quadratic_weighted_kappa"])
    display(diag)
    diag.to_csv(RESULTS/"annotation_reliability_diagnostic.csv",index=False)
    if diag.double_coded_replies.lt(diag.expected_replies).any(): warnings.warn("Independent coding is incomplete.")
else:
    print("Independent coding agreement is pending.")
if "context" in data and len(pairs):
    c=pairs.merge(data["context"],on="pair_id").loc[lambda x:x.missing.eq("yes")]
    queues=[list(c.loc[c.subset.eq(s),"pair_id"].sample(frac=1,random_state=SEED+i))
            for i,s in enumerate(SUBSETS)]
    from itertools import zip_longest
    sel=[key for group in zip_longest(*queues) for key in group if key is not None][:N_PROBE]
    draw=c.set_index("pair_id").loc[sel,["prompt","reply_a","reply_b","condition_note"]].reset_index()
    draw["order"]=rng(43).choice(["1_first","2_first"],len(draw))
    draw.to_csv(PRIVATE/"context_probe_packet.csv",index=False)
    if "probes" in data:
        p=data["probes"]
        valid=(set(p.pair_id)==set(sel) and p.order.eq(p.pair_id.map(draw.set_index("pair_id").order)).all()
               and p[["condition_1","condition_2"]].ne("").all().all()
               and p.condition_1.ne(p.condition_2).all())
        if not valid:
            data.pop("probes")
            pending["probes"]="returned probes do not match the seeded draw or documented conditions"
    print(f"RQ4 draw: {len(sel)} of {len(c)} eligible.")
```

### 6 · Define common scientific output helpers

Use the same styling and export conventions for all exhibits. Use a subset-stratified cluster bootstrap for paired estimates.

```python
def ready(rq,*need):
    missing=[k for k in need if k not in data]
    if not len(pairs): missing.insert(0,"frozen source sample")
    if missing:
        status[rq]="pending: "+", ".join(missing)
        print(rq,status[rq])
        return False
    status[rq]="computed"
    return True

def table(name,x):
    display(x)
    x.to_csv(RESULTS/f"{name}.csv",index=False)
def emit(name,fig):
    fig.tight_layout()
    fig.savefig(RESULTS/f"{name}.pdf",bbox_inches="tight")
    display(fig)
    plt.close(fig)
def roles(d):
    x=d.merge(pairs[["pair_id","subset","chosen_side","partition","cluster_id"]],on="pair_id",validate="many_to_one")
    x["preference"]=np.where(x.side.eq(x.chosen_side),"chosen","rejected")
    return x
def rate(n,d): return n/d if d else np.nan
def ratio_frame(n,d): return n/d.replace(0,np.nan)
def horiz(x,label,value,title,subtitle):
    fig,ax=plt.subplots(figsize=(9,max(3,1.9+.34*len(x))))
    y=np.arange(len(x))
    vals=x[value].astype(float).to_numpy()
    ax.barh(y,np.nan_to_num(vals),color=[COLORS[i%3] for i in y],height=.58)
    ax.set_yticks(y,x[label].str.replace("_"," "))
    ax.invert_yaxis()
    ax.set_xlim(0,1.18)
    ax.xaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
    for i,v in enumerate(vals):
        ax.annotate(f"{v:.1%}" if np.isfinite(v) else "No eligible cases",
                    (v if np.isfinite(v) else 0,i),xytext=(4,0),textcoords="offset points",va="center")
    ax.grid(axis="x",alpha=.2)
    ax.set_axisbelow(True)
    ax.set_title(title,loc="left",fontweight="bold")
    fig.text(.01,.01,subtitle,fontsize=8)
    fig.subplots_adjust(bottom=.18)
    return fig
def heat(x,title,xlab,ylab):
    fig,ax=plt.subplots(figsize=(7,4.6))
    vmax=max(1,x.to_numpy().max())
    im=ax.imshow(x,cmap="PuBu",vmin=0,vmax=vmax)
    ax.set_xticks(range(len(x.columns)),[s.replace("_"," ") for s in x.columns],rotation=20,ha="right")
    ax.set_yticks(range(len(x.index)),[s.replace("_"," ") for s in x.index])
    for i in range(len(x)):
        for j in range(len(x.columns)):
            n=int(x.iloc[i,j])
            ax.text(j,i,str(n),ha="center",va="center",color="white" if n>.6*vmax else "#142A41")
    ax.set(xlabel=xlab,ylabel=ylab,title=title)
    fig.colorbar(im,ax=ax,label="Count")
    return fig
def boots(frame,stat,step):
    if len(frame)<2 or frame.cluster_id.nunique()<2: return [np.nan,np.nan]
    f=frame.reset_index(drop=True)
    strata=[[x.index.to_numpy() for _,x in g.groupby("cluster_id",sort=False)]
            for _,g in f.groupby("subset",sort=True)]
    random=rng(step)
    reps=[]
    for _ in range(BOOT):
        index=np.concatenate([h[i] for h in strata for i in random.integers(0,len(h),len(h))])
        reps.append(stat(f.iloc[index]))
    values=np.array(reps)
    values=values[np.isfinite(values)]
    return list(np.quantile(values,[.025,.975])) if len(values) else [np.nan,np.nan]
```

### 7 · RQ1 value trade-offs

Compare ordinal scores within matched pairs. Count an observed conflict only when each reply wins on at least one jointly rateable dimension. Ordered crossings may occur more than once per pair.

```python
if ready("RQ1","values"):
    v=roles(data["values"])
    wins=pd.DataFrame(index=pairs.pair_id)
    for dim in DIMS:
        w=v.assign(score=pd.to_numeric(v[dim],errors="coerce")).pivot(
            index="pair_id",columns="preference",values="score")
        wins[dim]=np.sign(w.chosen-w.rejected)
    context=pairs.set_index("pair_id")[["subset"]]
    wins=wins.join(context)
    conflict=wins[DIMS].eq(1).any(axis=1)&wins[DIMS].eq(-1).any(axis=1)
    wins["outcome"]=np.select([conflict,wins[DIMS].isna().any(axis=1)],
                              ["opposed","unrateable"],default="no opposition")
    summary=pd.crosstab(wins.subset,wins.outcome).reindex(
        columns=["opposed","no opposition","unrateable"],fill_value=0)
    summary["pairs"]=summary.sum(axis=1)
    summary["opposed_share"]=ratio_frame(summary.opposed,summary.pairs)
    summary["unrateable_share"]=ratio_frame(summary.unrateable,summary.pairs)
    table("rq1_table1_pair_outcomes",summary.reset_index())
    crossing=pd.DataFrame([
        (s,a,b,int((g[a].eq(1)&g[b].eq(-1)).sum()),len(g))
        for s,g in wins.groupby("subset") for a in DIMS for b in DIMS if a!=b],
        columns=["subset","chosen_advantage","rejected_advantage","count","sample_pairs"])
    crossing["share_of_pairs"]=ratio_frame(crossing["count"],crossing.sample_pairs)
    table("rq1_table2_directional_crossings",crossing)
    chart=summary.reset_index().melt(id_vars="subset",value_vars=["opposed_share","unrateable_share"],
                                   var_name="outcome",value_name="rate")
    chart["group"]=chart.subset+" · "+chart.outcome.str.replace("_share","")
    emit("rq1_figure1_tradeoff_rates",horiz(chart,"group","rate","Observed value conflicts by subset",
         "Denominator: all sampled pairs in each subset. Unrateable means an incomplete comparison with no observed conflict."))
    fig_data=crossing.groupby(["chosen_advantage","rejected_advantage"])["count"].sum().unstack(fill_value=0)
    emit("rq1_figure2_direction",heat(fig_data,"Direction of observed value trade-offs",
         "Value favouring rejected reply","Value favouring chosen reply"))
```

### 8 · RQ2 source-checked factual confidence

Use claim-level source checks to construct a reply flag. Compare only pairs where both replies have checkable claims in the primary estimate, then show broader and stricter estimates.

```python
reply_flags=pd.DataFrame()
if ready("RQ2","coverage","claims"):
    inventory=data["coverage"]
    claims=roles(data["claims"])
    observed=set(zip(claims.pair_id,claims.side))
    expected=set(zip(inventory.loc[inventory.inventory.eq("complete"),"pair_id"],
                     inventory.loc[inventory.inventory.eq("complete"),"side"]))
    if observed!=expected:
        status["RQ2"]="pending: claim rows and complete reply inventories disagree"
        print(status["RQ2"])
    else:
        claims["checkable"]=claims.evidence.ne("uncheckable")
        base=claims.confidence.eq("categorical")&claims.consequential.eq("yes")
        claims["flag"]=base&claims.evidence.isin(["contradicted","insufficient"])
        claims["strict"]=base&claims.evidence.eq("contradicted")
        agg=claims.groupby(["pair_id","side"])[["checkable","flag","strict"]].any().reset_index()
        reply_flags=roles(inventory.merge(agg,on=["pair_id","side"],how="left"))
        for col in ["checkable","flag","strict"]:
            reply_flags[col]=reply_flags[col].eq(True)
        d=reply_flags
        meta=pairs.set_index("pair_id")[["subset","cluster_id"]]
        w=d.pivot(index="pair_id",columns="preference",values=["flag","strict","checkable","inventory"])
        eligible=w["checkable"].all(axis=1)
        broad=w["inventory"].ne("unassessed").all(axis=1)
        prim=w["flag"].loc[eligible].astype(int).join(meta)
        prim["gap"]=prim.chosen-prim.rejected
        prim["outcome"]=np.select([prim.chosen.eq(1)&prim.rejected.eq(1),prim.gap.eq(1),prim.gap.eq(-1)],
                                  ["both","chosen_only","rejected_only"],default="neither")
        counts=pd.crosstab(prim.subset,prim.outcome).reindex(index=SUBSETS,
                columns=["chosen_only","rejected_only","both","neither"],fill_value=0)
        counts["eligible"]=counts.sum(axis=1)
        counts["sampled"]=meta.groupby("subset").size()
        counts.loc["balanced overall"]=counts.sum()
        for i, label in enumerate(counts.index):
            subset_pairs=prim if label=="balanced overall" else prim.loc[prim.subset.eq(label)]
            if label=="balanced overall":
                interval=boots(subset_pairs,lambda x:x.gap.mean(),80)
            else:
                interval=boots(subset_pairs,lambda x:x.gap.mean(),85+i)
            counts.loc[label,"chosen_flag_rate"]=subset_pairs.chosen.mean()
            counts.loc[label,"rejected_flag_rate"]=subset_pairs.rejected.mean()
            counts.loc[label,"paired_gap_pp"]=100*subset_pairs.gap.mean()
            counts.loc[label,"ci_95_lower_pp"]=100*interval[0]
            counts.loc[label,"ci_95_upper_pp"]=100*interval[1]
        counts.index.name="subset"
        table("rq2_table1_matched_outcomes",counts.reset_index())
        outcomes=(claims.groupby(["preference","confidence","evidence"],dropna=False).size()
                  .rename("claims").reset_index())
        table("rq2_table2_claim_source_checks",outcomes)
        compare=[]
        for i,(label,f) in enumerate([("Primary: both checkable",prim),
            ("Strict: contradictions only",w["strict"].loc[eligible].astype(int).join(meta)),
            ("Broader: both assessed",w["flag"].loc[broad].astype(int).join(meta))]):
            f=f.assign(gap=f.chosen-f.rejected)
            lo,hi=boots(f,lambda x:x.gap.mean(),80+i)
            compare.append((label,len(f),f.gap.mean(),lo,hi))
        comp=pd.DataFrame(compare,columns=["contrast","pairs","estimate","lower","upper"])
        fig,ax=plt.subplots(figsize=(8,3.5))
        ax.axvline(0,color="#777777",linewidth=1)
        for i,r in comp.iterrows():
            if np.isfinite(r.estimate):
                if np.isfinite(r.lower): ax.hlines(i,r.lower,r.upper,color=COLORS[i],linewidth=2)
                ax.scatter(r.estimate,i,color=COLORS[i],zorder=3)
            else: ax.text(.02,i,"No eligible pairs",va="center")
        ax.set_yticks(range(len(comp)),[f"{r.contrast} (n={r.pairs})" for _,r in comp.iterrows()])
        ax.invert_yaxis()
        ax.xaxis.set_major_formatter(plt.matplotlib.ticker.FuncFormatter(lambda v,_: f"{v*100:g}"))
        ax.set(xlabel="Chosen minus rejected (percentage points)",title="Paired gap in unsupported confidence · 95% intervals")
        emit("rq2_figure1_paired_uncertainty",fig)
        if "values" in data:
            s=d.merge(data["values"][["pair_id","side","strategy"]],on=["pair_id","side"])
            s=s.loc[s.checkable].groupby(["strategy","preference"]).flag.agg(["mean","size"]).reset_index()
            s["label"]=s.strategy+" · "+s.preference+" · n="+s["size"].astype(str)
            if len(s):
                emit("rq2_figure2_strategies",horiz(s,"label","mean","Unsupported high confidence by strategy",
                     "Denominator: checkable replies in each group. Strategy comparisons are descriptive."))
            else:
                status["RQ2"]="partial: no checkable replies for strategy exhibit"
        else:
            status["RQ2"]="partial: strategy exhibit requires values.csv"
```

### 9 · RQ3 stakeholder treatment and missing context

Compare explicit, implied, absent, and unresolved recognition among coded applicable roles. For missing context, count each sampled pair once and expose unresolved reviews.

```python
if ready("RQ3","roles","context"):
    r=data["roles"]
    long=r.melt(id_vars=["pair_id","role_id","role_type"],value_vars=["visibility_A","visibility_B"],
                var_name="side",value_name="visibility")
    long.side=long.side.str[-1]
    long=roles(long)
    a=pd.crosstab([long.role_type,long.preference],long.visibility).reindex(columns=visibility,fill_value=0)
    a["applicable"]=a.sum(axis=1)
    table("rq3_table1_role_visibility",a.reset_index())
    b=pairs[["pair_id","subset"]].merge(data["context"],on="pair_id",validate="one_to_one")
    b["condition"]=np.where(b.missing.eq("yes"),b.condition_type,b.missing)
    c=b.groupby(["subset","condition"],sort=True).size().rename("pairs").reset_index()
    c["subset_total"]=c.subset.map(b.groupby("subset").size())
    c["share"]=ratio_frame(c.pairs,c.subset_total)
    table("rq3_table2_missing_conditions",c)
    fig,ax=plt.subplots(figsize=(8,4.2))
    left=np.zeros(len(a))
    for i,k in enumerate(visibility):
        vals=ratio_frame(a[k],a.applicable).fillna(0).to_numpy()
        ax.barh(range(len(a)),vals,left=left,color=COLORS[i],label=k)
        left+=vals
    ax.set_yticks(range(len(a)),[f"{x.replace('_',' ')} · {y} (n={n})"
                                for (x,y),n in a.applicable.items()])
    ax.invert_yaxis()
    ax.xaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
    ax.set(xlim=(0,1),xlabel="Share of coded applicable roles",title="Recognition by stakeholder and reply role")
    ax.legend(ncol=4,loc="upper center",bbox_to_anchor=(.5,-.18))
    emit("rq3_figure1_role_recognition",fig)
    miss=b.assign(miss=b.missing.eq("yes"),unknown=b.missing.eq("unresolved")).groupby("subset").agg(
        missing=("miss","mean"),unresolved=("unknown","mean")).reset_index()
    plot=miss.melt(id_vars="subset",var_name="class",value_name="rate")
    plot["label"]=plot.subset+" · "+plot["class"]
    emit("rq3_figure2_missing_context",horiz(plot,"label","rate","Missing stakeholder context",
         "Denominator: all sampled pairs in each subset; unresolved records shown separately."))
```

### 10 · RQ4 controlled context changes

Compare judgments across the two validated conditions. Report direct A↔B reversal separately and show a determinate-only sensitivity denominator.

```python
if ready("RQ4","probes"):
    original=data["probes"].merge(pairs[["pair_id","subset"]],on="pair_id")
    x=original.loc[original.validated.eq("yes")].copy()
    if x.empty:
        status["RQ4"]="pending: no independently validated context probes"
        print(status["RQ4"])
    else:
        mat=pd.crosstab(x.judgment_1,x.judgment_2).reindex(index=states,columns=states,fill_value=0)
        table("rq4_table1_transition_matrix",mat.rename_axis("condition_1").reset_index())
        direct=(x.judgment_1.eq("A")&x.judgment_2.eq("B"))|(x.judgment_1.eq("B")&x.judgment_2.eq("A"))
        x["change"]=np.select([x.judgment_1.eq(x.judgment_2),direct],["stable","direct reversal"],default="other change")
        x["determinate"]=x[["judgment_1","judgment_2"]].ne("indeterminate").all(axis=1)
        g=pd.crosstab(x.subset,x.change).reindex(index=SUBSETS,columns=["stable","direct reversal","other change"],fill_value=0)
        g["validated"]=g.sum(axis=1)
        g["assigned"]=original.groupby("subset").size().reindex(g.index,fill_value=0)
        g["both_determinate"]=x.groupby("subset").determinate.sum().reindex(g.index,fill_value=0)
        g["direct_reversal_among_determinate"]=ratio_frame(g["direct reversal"],g.both_determinate)
        table("rq4_table2_change_by_subset",g.reset_index())
        emit("rq4_figure1_transition_counts",heat(mat,"Judgment changes under one altered condition",
             "Judgment under condition 2","Judgment under condition 1"))
        chart=g[["direct reversal","other change"]].div(g.validated.replace(0,np.nan),axis=0)
        chart=chart.reset_index().melt(id_vars="subset",var_name="change",value_name="rate")
        chart["label"]=chart.subset+" · "+chart.change
        emit("rq4_figure2_composition",horiz(chart,"label","rate","Which judgments change?",
             "Denominator: independently validated probes in each subset; includes indeterminate judgments."))
```

### 11 · RQ5 held-out review triage

Compare two screening rules at the same 20% budget and display the recall curve as a descriptive sensitivity analysis. Unresolved reviews remain in the ranking pool and are shown explicitly.

```python
if ready("RQ5","values","context","reviews","coverage","claims"):
    eligible=reply_flags.loc[reply_flags.partition.eq("heldout")] if len(reply_flags) else pd.DataFrame()
    if eligible.empty or eligible.inventory.eq("unassessed").any():
        status["RQ5"]="pending: held-out claim inventories must be complete"
        print(status["RQ5"])
    else:
        h=pairs.loc[pairs.partition.eq("heldout"),["pair_id","subset","cluster_id"]].copy()
        h=h.merge(data["reviews"],on="pair_id",validate="one_to_one")
        h["refusal"]=h.pair_id.map(data["values"].assign(is_refusal=lambda f:f.strategy.eq("refusal"))
                                   .groupby("pair_id").is_refusal.any()).astype(int)
        h["evidence_context"]=(h.pair_id.map(eligible.groupby("pair_id").flag.any()).astype(int)
                               +h.pair_id.map(data["context"].set_index("pair_id").missing.eq("yes")).astype(int))
        h["tie_key"]=h.pair_id.map(lambda x:identity(str(SEED)+x))
        rules={"Refusal only":"refusal","Evidence and context":"evidence_context"}
        def rank(frame,rule,budget=.20):
            n=max(1,int(np.floor(len(frame)*budget)))
            return frame.sort_values([rule,"tie_key"],ascending=[False,True],kind="stable").head(n)
        def metrics(frame,rule,budget=.20,unresolved=None):
            q=rank(frame,rule,budget)
            ref=frame.needs_review if unresolved is None else frame.needs_review.replace("unresolved",unresolved)
            sel=q.needs_review if unresolved is None else q.needs_review.replace("unresolved",unresolved)
            found=int(sel.eq("yes").sum())
            positives=int(ref.eq("yes").sum())
            return {"reviewed":len(q),"resolved_reviewed":int(sel.ne("unresolved").sum()),
                    "unresolved_reviewed":int(q.needs_review.eq("unresolved").sum()),"reference_cases":positives,
                    "found":found,"missed":positives-found,"recall":rate(found,positives),
                    "precision":rate(found,int(sel.ne("unresolved").sum()))}
        stats=[]
        for label,col in rules.items():
            m=metrics(h,col)
            lo,hi=boots(h,lambda v:metrics(v,col)["recall"],105)
            pl,ph=boots(h,lambda v:metrics(v,col)["precision"],105)
            stats.append({"rule":label,**m,"recall_low":lo,"recall_high":hi,"precision_low":pl,
                          "precision_high":ph,"heldout_pairs":len(h),
                          "unresolved_reference":int(h.needs_review.eq("unresolved").sum()),
                          "recall_if_all_unresolved_positive":metrics(h,col,unresolved="yes")["recall"],
                          "recall_if_all_unresolved_negative":metrics(h,col,unresolved="no")["recall"]})
        summary=pd.DataFrame(stats)
        def delta(v): return metrics(v,"evidence_context")["recall"]-metrics(v,"refusal")["recall"]
        lo,hi=boots(h,delta,106)
        summary["paired_recall_gain"]=delta(h)
        summary["paired_gain_low"]=lo
        summary["paired_gain_high"]=hi
        table("rq5_table1_heldout_performance",summary)
        positives=h.loc[h.needs_review.eq("yes"),"pair_id"]
        a=set(rank(h,"refusal").pair_id)
        b=set(rank(h,"evidence_context").pair_id)
        overlap=pd.Series([("both" if x in a and x in b else "refusal_only" if x in a else
                            "evidence_context_only" if x in b else "neither") for x in positives],dtype=str)
        counts=overlap.value_counts().reindex(["both","refusal_only","evidence_context_only","neither"],
                                              fill_value=0).rename_axis("detection").reset_index(name="cases")
        counts["reference_cases"]=len(positives)
        table("rq5_table2_detection_overlap",counts)
        fig,ax=plt.subplots(figsize=(7.8,4.4))
        for i,metric in enumerate(["recall","precision"]):
            for j,r in summary.iterrows():
                x=i+(-.11 if j==0 else .11)
                if np.isfinite(r[metric]):
                    l,u=r[f"{metric}_low"],r[f"{metric}_high"]
                    if np.isfinite(l): ax.vlines(x,l,u,color=COLORS[j],linewidth=2)
                    ax.scatter(x,r[metric],color=COLORS[j],label=r.rule if i==0 else None,zorder=3)
        ax.set(xticks=[0,1],xticklabels=["Recall","Precision"],xlim=(-.5,1.5),ylim=(-.04,1.04),
               title="Held-out review utility at 20% capacity · 95% intervals")
        ax.yaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
        if ax.get_legend_handles_labels()[0]: ax.legend()
        if not len(positives): ax.text(.5,.12,"No adjudicated positive cases; recall undefined",ha="center")
        emit("rq5_figure1_matched_budget",fig)
        fig,ax=plt.subplots(figsize=(7.8,4.4))
        budgets=np.arange(.05,.55,.05)
        for j,(label,col) in enumerate(rules.items()):
            ax.plot(budgets,[metrics(h,col,b)["recall"] for b in budgets],marker="o",color=COLORS[j],label=label)
        ax.axvline(.20,color="#555555",linestyle="--",label="Primary budget")
        ax.set(xlabel="Share of held-out pairs sent to review",ylabel="Recall among reference cases",ylim=(-.03,1.03),
               title="Hold-out review yield across review budgets")
        for axis in (ax.xaxis,ax.yaxis): axis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
        ax.legend()
        emit("rq5_figure2_review_budget_curve",fig)
```
